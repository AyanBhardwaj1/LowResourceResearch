```python
!wget -O bewiki.xml.bz2 https://dumps.wikimedia.org/bewiki/latest/bewiki-latest-pages-articles.xml.bz2

import bz2
import re

def extract_articles(file_path, max_articles=100):
    articles = []
    with bz2.open(file_path, 'rt', encoding='utf-8', errors='ignore') as f:
        in_page = False
        title = text = ''
        for line in f:
            if '<page>' in line:
                in_page = True
                title, text = '', ''
            elif '</page>' in line and in_page:
                if title and text and not title.startswith(('Шаблон', 'File:', 'Category:')):
                    articles.append({'title': title, 'text': text.strip()})
                in_page = False
                if len(articles) >= max_articles:
                    break
            elif in_page:
                if '<title>' in line:
                    title = re.sub(r'<.*?>', '', line).strip()
                elif '<text' in line:
                    text = re.sub(r'<.*?>', '', line)
                elif text:
                    text += line
    return articles

articles = extract_articles('bewiki.xml.bz2', max_articles=5)

for i, a in enumerate(articles):
    print(f"\n🔹 Article {i+1}: {a['title']}\n{'-'*80}\n{a['text'][:500]}...\n{'='*80}")
```

```python
# Belarusian Wikipedia Dump Scraper for NLP Dataset Creation

"""
This script is designed for Google Colab, but will also work locally with minor modifications. It does the following:

- Scrapes the index page for Belarusian Wikipedia dumps.
- Downloads all relevant dump files ("pages-articles" XMLs, multistream, etc.).
- Extracts plain text from the XML dumps using WikiExtractor.
- Outputs a dataset (e.g., one article per line, JSONL or TXT) ready for NLP/LLM tasks.

Instructions: Just run each cell in order. Internet required.
"""

# --- 1. Install dependencies ---
!pip install wikiextractor tqdm requests

import os
import requests
from bs4 import BeautifulSoup
from tqdm import tqdm

# --- 2. Set the dump page URL ---
BEWIKI_DUMP_URL = 'https://dumps.wikimedia.org/bewiki/latest/'
OUTPUT_DIR = 'bewiki_dumps'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# --- 3. Scrape the dump index for all files ---
resp = requests.get(BEWIKI_DUMP_URL)
soup = BeautifulSoup(resp.text, 'html.parser')

dump_files = []
for link in soup.find_all('a'):
    href = link.get('href')
    if href and any(x in href for x in ['pages-articles', 'xml.bz2']):
        dump_files.append(BEWIKI_DUMP_URL + href)

print(f"Found {len(dump_files)} dump files:")
for f in dump_files:
    print(f)

# --- 4. Download all the dump files ---
for url in tqdm(dump_files, desc="Downloading dumps"):
    filename = os.path.join(OUTPUT_DIR, url.split('/')[-1])
    if not os.path.exists(filename):
        r = requests.get(url, stream=True)
        with open(filename, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
    else:
        print(f"Already downloaded: {filename}")

# --- 5. Extract articles using WikiExtractor ---
# Note: This tool extracts plain text from Wikipedia XML dumps.

from wikiextractor.WikiExtractor import main as wikiextractor_main

# We'll use the largest/most recent pages-articles XML as the main dump
main_xml = None
for f in os.listdir(OUTPUT_DIR):
    if 'pages-articles' in f and f.endswith('.xml.bz2'):
        main_xml = os.path.join(OUTPUT_DIR, f)
        break

assert main_xml, "No 'pages-articles' XML dump found!"

EXTRACT_DIR = 'bewiki_extracted'
os.makedirs(EXTRACT_DIR, exist_ok=True)

# Run WikiExtractor
!python3 -m wikiextractor.WikiExtractor {main_xml} --json --output {EXTRACT_DIR} --no-templates --processes 2

# --- 6. Combine all extracted files into a single JSONL dataset ---
import json
dataset_file = 'belarusian_wikipedia.jsonl'

with open(dataset_file, 'w', encoding='utf-8') as fout:
    for root, dirs, files in os.walk(EXTRACT_DIR):
        for fname in files:
            if fname.endswith('.json'):
                with open(os.path.join(root, fname), encoding='utf-8') as fin:
                    for line in fin:
                        fout.write(line)

print(f"✅ Belarusian Wikipedia dataset saved as: {dataset_file}")
```
